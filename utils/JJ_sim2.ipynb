```python
# Re-importing necessary libraries and re-defining parameters due to code reset
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint

# Parameters
Ic = 1e-6  # Critical current (A)
R = 1  # Shunt resistance (Ohm)
C = 1e-8  # Shunt capacitance (F)
phi0 = 2.067e-15  # Magnetic flux quantum (Wb)
hbar_over_2e = phi0 / (2 * np.pi)  # Reduced flux quantum
t = np.linspace(0, 5e-3, 1000)  # Time array for numerical integration

# Define the RCSJ differential equation
def rcsj_ode(phi_dot, t, I_ext):
    phi, phi_t = phi_dot
    V = hbar_over_2e * phi_t
    dphi_dt = phi_t
    dphi_t_dt = (I_ext - Ic * np.sin(phi) - V / R) / (C * hbar_over_2e)
    return [dphi_dt, dphi_t_dt]

# Simulate the RCSJ IV curve by sweeping current from -2 µA to 2 µA
currents_applied = np.linspace(-2e-6, 2e-6, 501)  # Applied current from -2 µA to 2 µA
voltages_measured = []

# Solve the RCSJ equation for each applied current
for I_ext in currents_applied:
    initial_conditions = [0, 0]
    sol = odeint(rcsj_ode, initial_conditions, t, args=(I_ext,))
    phi_t = sol[:, 1]
    V_junction = hbar_over_2e * phi_t
    voltages_measured.append(np.mean(V_junction[-100:]))

# Convert voltages to an array for plotting
voltages_measured = np.array(voltages_measured)

# Plot the RCSJ response for applied current from -1A to 1A
plt.figure(figsize=(8, 6))
plt.plot(currents_applied, voltages_measured, label='RCSJ IV Curve (Current Sweep)', color='red')
plt.xlabel('Applied Current (A)')
plt.ylabel('Voltage (V)')
plt.title('RCSJ IV Curve - Current Sweep from -1A to 1A')
plt.grid(True)
plt.legend()
plt.show()

```

```python
from scipy.integrate import odeint
import numpy as np
import scipy.constants as const
from matplotlib import pyplot as plt
from scipy.interpolate import interp1d
from scipy.integrate import quad
import collections
```

```python
Ic = 1e-6
Rsg = 10e3
Rn = Rsg
Cj = 2e-15
Rin = 100e0
Vgap = 4*Ic*Rn/np.pi #Gap voltage linked to IcRn, SIS relation
Vn = 1.1 #0.1*Vgap/Ic/Rin
V0 = Vn*Rin*Ic
phi0 = const.value('mag. flux quantum')
wp = 1/np.sqrt(Cj*phi0/2/np.pi/Ic)

params = {'Vn':Vn, 'Rsg': Rsg, 'Rn':Rn, 'Vgap':Vgap, 'Cj':Cj, 'Rin':Rin, 'wp':wp}
print(Vn)
```

```python
taustop = 20* wp * Rsg*Cj
params['taustop'] = taustop
print(params)
```

```python
def Rj(V,params):
    Vgap = params['Vgap']
    if V>Vgap:
        return params['Rn']
    elif V<=Vgap:
        return params['Rsg']

def Q(V,params):
    Cj = params['Cj']
    Rin = params['Rin']
    wp = params['wp']
    return Cj*wp/(1/Rj(V,params)+1/Rin)

def Vs(t,params):
    if not hasattr(t,'__len__'):
        t = np.array([t])
    result = []
    for x in t:
        if x <params['taustop']/2:
            result.append(x/(params['taustop']/2)*params['Vn'])
        else:
            result.append(params['Vn'])
    #return params['Vn']
    result = np.ones(len(t))*params['Vn']
    return np.array(result)

def VJJmodel(z,t,params):
    f = z[0]
    g = z[1]
    V = phi0/2/np.pi*g
    fp = g
    gp = Vs(t,params)-(1/Q(V,params)*g+np.sin(f))
    return np.array([fp,gp])
```

```python
ts = np.linspace(0,2*taustop,2000)
sols = odeint(VJJmodel,[0,0],ts, args=tuple([params]) )
solsfunc = interp1d(ts,sols.T)
Vjunc = lambda x: phi0/2/np.pi * wp * solsfunc(x)[1]
taurep = 2*np.pi/np.average(np.diff(sols[:,0])/np.diff(ts))
```

```python
fig = plt.figure(figsize=(10,10))
plt.clf()
plt.subplot(3, 1, 1)
plt.title('Voltage/Vgap')
plt.plot(ts,Vjunc(ts) / Vgap)
plt.subplot(3, 1, 2)
plt.title('Current')
plt.plot(ts,np.sin(solsfunc(ts)[0]))
plt.plot(ts,(Vs(ts,params)*Ic*Rin-Vjunc(ts))/Rin/Ic)
plt.ylim([-2,2])
plt.subplot(3, 1, 3)
plt.title('Phase')
plt.plot(ts,solsfunc(ts)[0])
plt.show()
plt.close(fig)
```
