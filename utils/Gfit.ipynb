```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import find_peaks, peak_prominences, peak_widths
from sklearn.mixture import GaussianMixture

np.random.seed(0)  

x_data = np.linspace(400, 700, 300)  
y_data = np.exp(-0.005*(x_data-500)**2) + np.exp(-0.005*(x_data-600)**2) + np.random.normal(0, 0.02, x_data.size) + 0.1*(x_data-400)**2/300**2

# 使用高斯混合模型來找到兩個峰值
gmm = GaussianMixture(n_components=2)
gmm.fit(np.vstack((x_data, y_data)).T)
means = gmm.means_[:, 0]
covariances = gmm.covariances_[:, 0, 0]
weights = gmm.weights_

# 計算擬合的高斯曲線
fit_y_data = np.zeros_like(y_data)
for mean, cov, weight in zip(means, covariances, weights):
    fit_y_data += weight * np.exp(-0.5 * ((x_data - mean) ** 2) / cov) / np.sqrt(2 * np.pi * cov)

# 顯示最顯著的兩個 peak 的波長
print("Most prominent peak wavelengths (GMM):", means)

fig, axs = plt.subplots(2, 1, figsize=(6, 8), sharex=True)
axs[0].plot(x_data, y_data, label='Original Data')
axs[0].plot(x_data, fit_y_data, label='GMM Fit', linestyle='--')
axs[0].set_title('Original Spectral Data with GMM Fit')
axs[0].legend()

axs[1].plot(x_data, y_data - fit_y_data, color='orange', label='Corrected Data')
axs[1].set_title('Baseline Corrected Data')
axs[1].legend()

for ax in axs:
    ax.set_xlabel('Wavelength (nm)')
    ax.set_ylabel('Intensity')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture

# Generate data
x_data = np.linspace(400, 700, 300)
y_data = (np.exp(-0.005*(x_data-500)**2) + 
          np.exp(-0.005*(x_data-600)**2) + 
          np.random.normal(0, 0.02, x_data.size) + 
          0.1*(x_data-400)**2/300**2)

# Combine x_data and y_data for fitting
data = np.vstack((x_data, y_data)).T

# Fit Gaussian Mixture Model (GMM) to the data
gmm = GaussianMixture(n_components=2, random_state=0)
gmm.fit(data)

# Get GMM results
means = gmm.means_[:, 1]
stds = np.sqrt(gmm.covariances_[:, 1, 1])
weights = gmm.weights_

# Generate Gaussian curves for visualization
fit_y = sum(w * np.exp(-0.5 * ((x_data - m) / s) ** 2) / (s * np.sqrt(2 * np.pi))
            for w, m, s in zip(weights, means, stds))

# Plot results
plt.figure(figsize=(10, 6))
plt.plot(x_data, y_data, 'k.', label='Data')
plt.plot(x_data, fit_y, 'r-', label='GMM Fit')
for mean, std in zip(means, stds):
    plt.axvline(mean, color='blue', linestyle='--')
plt.xlabel('X Data')
plt.ylabel('Y Data')
plt.legend()
plt.title('Gaussian Mixture Fit with Two Peaks')
plt.show()

# Print the results
for i, (mean, std, weight) in enumerate(zip(means, stds, weights)):
    print(f"Peak {i+1}: Mean = {mean:.2f}, Std Dev = {std:.2f}, Weight = {weight:.2f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.mixture import GaussianMixture

# Generate data
x_data = np.linspace(400, 700, 300)
y_data = (np.exp(-0.005*(x_data-500)**2) + 
          np.exp(-0.005*(x_data-600)**2) + 
          np.random.normal(0, 0.02, x_data.size) + 
          0.1*(x_data-400)**2/300**2)

# Fit Gaussian Mixture Model (GMM) on x_data weighted by y_data
gmm = GaussianMixture(n_components=2, random_state=0)
gmm.fit(x_data.reshape(-1, 1), sample_weight=y_data)

# Extract GMM parameters
means = gmm.means_.flatten()
stds = np.sqrt(gmm.covariances_).flatten()
weights = gmm.weights_

# Generate individual Gaussian curves
gaussian_peaks = [
    w * np.exp(-0.5 * ((x_data - m) / s)**2) / (s * np.sqrt(2 * np.pi))
    for w, m, s in zip(weights, means, stds)
]

# Sum of both Gaussians for the total fit
fit_y = sum(gaussian_peaks)

# Plot the data and fits
plt.figure(figsize=(10, 6))
plt.plot(x_data, y_data, 'k.', label='Data')
plt.plot(x_data, fit_y, 'r-', label='Total GMM Fit')
plt.plot(x_data, gaussian_peaks[0], 'g--', label='Gaussian Peak 1')
plt.plot(x_data, gaussian_peaks[1], 'b--', label='Gaussian Peak 2')
for mean, std in zip(means, stds):
    plt.axvline(mean, color='blue', linestyle='--')
plt.xlabel('X Data')
plt.ylabel('Y Data')
plt.legend()
plt.title('Gaussian Mixture Fit with Two Peaks')
plt.show()

# Print the results
for i, (mean, std, weight) in enumerate(zip(means, stds, weights)):
    print(f"Peak {i+1}: Mean = {mean:.2f}, Std Dev = {std:.2f}, Weight = {weight:.2f}")
```

```python
import plotly.graph_objects as go
import numpy as np
from sklearn.mixture import GaussianMixture
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import find_peaks, peak_prominences, peak_widths

np.random.seed(0)  

x_data = np.linspace(400, 700, 300)  
y_data = np.exp(-0.005*(x_data-500)**2) + np.exp(-0.005*(x_data-600)**2) + np.random.normal(0, 0.02, x_data.size) + 0.1*(x_data-400)**2/300**2

# 調整 distance 和 prominence 參數
peaks, _ = find_peaks(y_data, distance=20, prominence=0.05)

# 計算每個 peak 的顯著性
prominences = peak_prominences(y_data, peaks)[0]

# 找出顯著性最高的兩個 peak
top_two_indices = np.argsort(prominences)[-2:]
top_two_peaks = peaks[top_two_indices]

widths = peak_widths(y_data, top_two_peaks, rel_height=0.5)

mask = np.ones_like(y_data, dtype=bool)

for peak, width in zip(top_two_peaks, widths[0]):
    mask[int(peak - width):int(peak + width)] = False

baseline_values = np.polyfit(x_data[mask], y_data[mask], 2)
baseline_poly = np.poly1d(baseline_values)
baseline = baseline_poly(x_data)
y_corrected = y_data - baseline

# 顯示最顯著的兩個 peak 的波長
top_two_wavelengths = x_data[top_two_peaks]
print("Most prominent peak wavelengths:", top_two_wavelengths)

fig, axs = plt.subplots(2, 1, figsize=(6, 8), sharex=True)
axs[0].plot(x_data, y_data, label='Original Data')
axs[0].plot(x_data, baseline, label='Baseline', linestyle='--')
axs[0].set_title('Original Spectral Data with Baseline')
axs[0].legend()

axs[1].plot(x_data, y_corrected, color='orange', label='Corrected Data')
axs[1].set_title('Baseline Corrected Data')
axs[1].legend()

for ax in axs:
    ax.set_xlabel('Wavelength (nm)')
    ax.set_ylabel('Intensity')

plt.tight_layout()
plt.show()
# Generate data
x_data = np.linspace(400, 700, 300)
y_data = (np.exp(-0.005*(x_data-500)**2) + 
          np.exp(-0.005*(x_data-600)**2) + 
          np.random.normal(0, 0.02, x_data.size) + 
          0.1*(x_data-400)**2/300**2)

# Ensure repetition counts are non-negative and at least 1
repetition_counts = np.clip((y_corrected / y_corrected.max() * 100).astype(int), 1, None)

# Expand data using repetition counts
weighted_x_data = np.repeat(x_data, repetition_counts)

# Fit Gaussian Mixture Model (GMM) with more components
gmm = GaussianMixture(n_components=2, random_state=0)
gmm.fit(weighted_x_data.reshape(-1, 1))

# Extract GMM parameters
means = gmm.means_.flatten()
stds = np.sqrt(gmm.covariances_).flatten()
weights = gmm.weights_

# Generate individual Gaussian curves
gaussian_peaks = [
    w * np.exp(-0.5 * ((x_data - m) / s)**2) / (s * np.sqrt(2 * np.pi))
    for w, m, s in zip(weights, means, stds)
]

# Sum of all Gaussians for the total fit
fit_y = sum(gaussian_peaks)

# Create the figure
fig = go.Figure()

# Add original data
fig.add_trace(go.Scatter(x=x_data, y=y_data, mode='markers', name='Data'))

# Add total GMM fit on secondary y-axis
fig.add_trace(go.Scatter(x=x_data, y=fit_y, mode='lines', name='Total GMM Fit', yaxis='y2'))

# Add individual Gaussian peaks on secondary y-axis
for i, peak in enumerate(gaussian_peaks):
    fig.add_trace(go.Scatter(x=x_data, y=peak, mode='lines', name=f'Gaussian Peak {i+1}', line=dict(dash='dash'), yaxis='y2'))

# Add vertical lines for means
for mean in means:
    fig.add_trace(go.Scatter(x=[mean, mean], y=[0, max(y_data)], mode='lines', name=f'Mean {mean:.2f}', line=dict(color='blue', dash='dash')))

# Update layout
fig.update_layout(
    title='Gaussian Mixture Fit with Three Peaks',
    xaxis_title='X Data',
    yaxis_title='Y Data',
    yaxis2=dict(title='GMM Fit and Peaks', overlaying='y', side='right')
)

# Show the figure
fig.show()

# Print the results
for i, (mean, std, weight) in enumerate(zip(means, stds, weights)):
    print(f"Peak {i+1}: Mean = {mean:.2f}, Std Dev = {std:.2f}, Weight = {weight:.2f}")
    # Perform baseline correction
    baseline_values = np.polyfit(x_data, y_data, 2)
    baseline_poly = np.poly1d(baseline_values)
    baseline = baseline_poly(x_data)
    y_corrected = y_data - baseline

    # Ensure repetition counts are non-negative and at least 1
    repetition_counts = np.clip((y_corrected / y_corrected.max() * 100).astype(int), 1, None)

    # Expand data using repetition counts
    weighted_x_data = np.repeat(x_data, repetition_counts)

    # Fit Gaussian Mixture Model (GMM) with more components
    gmm = GaussianMixture(n_components=3, random_state=0)
    gmm.fit(weighted_x_data.reshape(-1, 1))

    # Extract GMM parameters
    means = gmm.means_.flatten()
    stds = np.sqrt(gmm.covariances_).flatten()
    weights = gmm.weights_

    # Generate individual Gaussian curves
    gaussian_peaks = [
        w * np.exp(-0.5 * ((x_data - m) / s)**2) / (s * np.sqrt(2 * np.pi))
        for w, m, s in zip(weights, means, stds)
    ]

    # Sum of all Gaussians for the total fit
    fit_y = sum(gaussian_peaks)

    # Create the figure
    fig = go.Figure()

    # Add original data
    fig.add_trace(go.Scatter(x=x_data, y=y_corrected, mode='markers', name='Data'))

    # Add total GMM fit on secondary y-axis
    fig.add_trace(go.Scatter(x=x_data, y=fit_y, mode='lines', name='Total GMM Fit', yaxis='y2'))

    # Add individual Gaussian peaks on secondary y-axis
    for i, peak in enumerate(gaussian_peaks):
        fig.add_trace(go.Scatter(x=x_data, y=peak, mode='lines', name=f'Gaussian Peak {i+1}', line=dict(dash='dash'), yaxis='y2'))

    # Add vertical lines for means
    for mean in means:
        fig.add_trace(go.Scatter(x=[mean, mean], y=[0, max(y_data)], mode='lines', name=f'Mean {mean:.2f}', line=dict(color='blue', dash='dash')))

    # Update layout
    fig.update_layout(
        title='Gaussian Mixture Fit with Three Peaks',
        xaxis_title='X Data',
        yaxis_title='Y Data',
        yaxis2=dict(title='GMM Fit and Peaks', overlaying='y', side='right')
    )

    # Show the figure
    fig.show()

    # Print the results
    for i, (mean, std, weight) in enumerate(zip(means, stds, weights)):
        print(f"Peak {i+1}: Mean = {mean:.2f}, Std Dev = {std:.2f}, Weight = {weight:.2f}")
```
